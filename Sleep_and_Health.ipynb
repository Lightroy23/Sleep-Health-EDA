## 7 Exploratory Data Analysis (EDA)

### 7.1 Distribution of Continuous Variables Variables using Histograms 

Histograms are used to examine the distribution of continuous numerical variables. They help identify the shape of the distribution, the concentration of values, possible skewness, and potential outliers.

The following histograms show the distributions of:
- Age
- Sleep Duration
- Heart Rate
- Daily Steps

```python
# =====================================================
# Distribution of Continuous Variables
# Histograms with Mean, Median and Skewness
# =====================================================

# Create a function to classify skewness
def skew_label(skew_value):
    if skew_value > 0.5:
        return "Positively Skewed"
    elif skew_value < -0.5:
        return "Negatively Skewed"
    else:
        return "Approximately Symmetric"


# Create a figure with 2 rows and 2 columns
fig, axes = plt.subplots(2, 2, figsize=(16, 11))

# =====================================================
# Histogram 1 : Age
# =====================================================

mean_age = df_sleep['Age'].mean()
median_age = df_sleep['Age'].median()
skew_age = df_sleep['Age'].skew()

axes[0,0].hist(df_sleep['Age'],
               bins=8,
               edgecolor='black')

# Mean line
axes[0,0].axvline(mean_age,
                  color='red',
                  linestyle='--',
                  linewidth=2,
                  label=f'Mean = {mean_age:.2f}')

# Median line
axes[0,0].axvline(median_age,
                  color='green',
                  linestyle='-',
                  linewidth=2,
                  label=f'Median = {median_age:.2f}')

axes[0,0].set_title('Age Distribution', fontsize=14, fontweight='bold')
axes[0,0].set_xlabel('Age (Years)')
axes[0,0].set_ylabel('Frequency')

# Display skewness information inside the plot
axes[0,0].text(
    0.02,
    0.95,
    f"Skewness = {skew_age:.2f}\n{skew_label(skew_age)}",
    transform=axes[0,0].transAxes,
    fontsize=10,
    verticalalignment='top',
    bbox=dict(facecolor='white', alpha=0.8)
)

axes[0,0].legend()

# =====================================================
# Histogram 2 : Sleep Duration
# =====================================================

mean_sleep = df_sleep['Sleep Duration(hours)'].mean()
median_sleep = df_sleep['Sleep Duration(hours)'].median()
skew_sleep = df_sleep['Sleep Duration(hours)'].skew()

axes[0,1].hist(df_sleep['Sleep Duration(hours)'],
               bins=8,
               edgecolor='black')

axes[0,1].axvline(mean_sleep,
                  color='red',
                  linestyle='--',
                  linewidth=2,
                  label=f'Mean = {mean_sleep:.2f}')

axes[0,1].axvline(median_sleep,
                  color='green',
                  linestyle='-',
                  linewidth=2,
                  label=f'Median = {median_sleep:.2f}')

axes[0,1].set_title('Sleep Duration Distribution', fontsize=14, fontweight='bold')
axes[0,1].set_xlabel('Sleep Duration (Hours)')
axes[0,1].set_ylabel('Frequency')

axes[0,1].text(
    0.02,
    0.95,
    f"Skewness = {skew_sleep:.2f}\n{skew_label(skew_sleep)}",
    transform=axes[0,1].transAxes,
    fontsize=10,
    verticalalignment='top',
    bbox=dict(facecolor='white', alpha=0.8)
)

axes[0,1].legend()

# =====================================================
# Histogram 3 : Heart Rate
# =====================================================

mean_hr = df_sleep['Heart Rate'].mean()
median_hr = df_sleep['Heart Rate'].median()
skew_hr = df_sleep['Heart Rate'].skew()

axes[1,0].hist(df_sleep['Heart Rate'],
               bins=8,
               edgecolor='black')

axes[1,0].axvline(mean_hr,
                  color='red',
                  linestyle='--',
                  linewidth=2,
                  label=f'Mean = {mean_hr:.2f}')

axes[1,0].axvline(median_hr,
                  color='green',
                  linestyle='-',
                  linewidth=2,
                  label=f'Median = {median_hr:.2f}')

axes[1,0].set_title('Heart Rate Distribution', fontsize=14, fontweight='bold')
axes[1,0].set_xlabel('Heart Rate (beats/minute)')
axes[1,0].set_ylabel('Frequency')

axes[1,0].text(
    0.02,
    0.95,
    f"Skewness = {skew_hr:.2f}\n{skew_label(skew_hr)}",
    transform=axes[1,0].transAxes,
    fontsize=10,
    verticalalignment='top',
    bbox=dict(facecolor='white', alpha=0.8)
)

axes[1,0].legend()

# =====================================================
# Histogram 4 : Daily Steps
# =====================================================

mean_steps = df_sleep['Daily Steps'].mean()
median_steps = df_sleep['Daily Steps'].median()
skew_steps = df_sleep['Daily Steps'].skew()

axes[1,1].hist(df_sleep['Daily Steps'],
               bins=8,
               edgecolor='black')

axes[1,1].axvline(mean_steps,
                  color='red',
                  linestyle='--',
                  linewidth=2,
                  label=f'Mean = {mean_steps:.0f}')

axes[1,1].axvline(median_steps,
                  color='green',
                  linestyle='-',
                  linewidth=2,
                  label=f'Median = {median_steps:.0f}')

axes[1,1].set_title('Daily Steps Distribution', fontsize=14, fontweight='bold')
axes[1,1].set_xlabel('Daily Steps')
axes[1,1].set_ylabel('Frequency')

axes[1,1].text(
    0.02,
    0.95,
    f"Skewness = {skew_steps:.2f}\n{skew_label(skew_steps)}",
    transform=axes[1,1].transAxes,
    fontsize=10,
    verticalalignment='top',
    bbox=dict(facecolor='white', alpha=0.8)
)

axes[1,1].legend()

# Improve spacing between plots
plt.tight_layout()

# Display the figure
plt.show()
```

#### 7.1.1 Interpretation of Histograms

The histograms provided an overview of the distribution of the continuous variables in the dataset. Including the mean, median, and skewness values made it easier to understand each distribution's shape, identify asymmetry, and determine whether the data is approximately normally distributed.

The histogram of age indicates that the participants range from approximately 27 to 59 years, with the highest concentration falling between 35 and 50 years.
```python
# =====================================================
# 7.4.1 Influence of Sleep Disorder on Health Indicators
# =====================================================

# Define a consistent colour palette
palette = {
    "None": "#66c2a5",          # Green
    "Insomnia": "#fc8d62",      # Orange
    "Sleep Apnea": "#8da0cb"    # Blue
}

# -----------------------------------------------------
# Function to create a boxplot with jitter
# -----------------------------------------------------

def plot_box_with_jitter(ax, y, title, ylabel):
    """
    Creates a boxplot with:
    - Different colours for each Sleep Disorder category
    - Mean displayed as a red diamond
    - Individual observations shown with jitter
    """

    # Boxplot
    sns.boxplot(
    data=df_sleep,
    x='Sleep Disorder',
    y=y,
    hue='Sleep Disorder',
    palette=palette,
    legend=False,
    width=0.6,
    linewidth=1.2,
    showmeans=True,
    meanprops={
        "marker": "D",
        "markerfacecolor": "red",
        "markeredgecolor": "black",
        "markersize": 6
    },
    ax=ax
)

    # Add jittered observations
    sns.stripplot(
        data=df_sleep,
        x='Sleep Disorder',
        y=y,
        color='black',
        jitter=0.25,
        alpha=0.45,
        size=3,
        ax=ax
    )

    # Formatting
    ax.set_title(title, fontsize=12, fontweight='bold')
    ax.set_xlabel('Sleep Disorder')
    ax.set_ylabel(ylabel)
   


# =====================================================
# Create a 1 × 4 Dashboard
# =====================================================

fig, axes = plt.subplots(1, 4, figsize=(24, 6))

plot_box_with_jitter(
    axes[0],
    'Quality of Sleep',
    'Sleep Quality',
    'Quality of Sleep'
)

plot_box_with_jitter(
    axes[1],
    'Sleep Duration(hours)',
    'Sleep Duration',
    'Sleep Duration (Hours)'
)

plot_box_with_jitter(
    axes[2],
    'Stress Level',
    'Stress Level',
    'Stress Level'
)

plot_box_with_jitter(
    axes[3],
    'Physical Activity Level',
    'Physical Activity',
    'Physical Activity Level'
)

# Overall title
fig.suptitle(
    'Influence of Sleep Disorder on Health Indicators',
    fontsize=18,
    fontweight='bold'
)

plt.tight_layout(rect=[0, 0, 1, 0.95])

plt.show()
```

#### Interpretation of Sleep Quality by Sleep Disorder

The boxplot indicates noticeable differences in sleep quality across the three sleep disorder categories. Participants with no sleep disorder generally report higher sleep quality scores, with a median of approximately 8 and relatively little variability. Participants with insomnia exhibit lower sleep quality, with scores concentrated around 6 to 7, indicating poorer perceived sleep. Individuals with sleep apnea display the widest variation in sleep quality, ranging from low to high scores, suggesting greater inconsistency in their sleep experiences. The mean values closely follow the medians in all three groups, indicating that extreme observations have limited influence on the central tendency.

#### Interpretation of Sleep Duration by Sleep Disorder

The distribution of sleep duration differs considerably among the three groups. Participants without a sleep disorder generally obtain longer sleep durations, with most observations clustered between 7 and 8 hours. Individuals with insomnia consistently record the shortest sleep durations, with values concentrated around 6.5 hours, indicating reduced nightly sleep. Participants with sleep apnea display the greatest variability in sleep duration, with observations spread across a wide range from approximately 6 to over 8 hours. This suggests that sleep apnea is associated with more inconsistent sleep patterns than the other groups.

#### Interpretation of Stress Level by Sleep Disorder

Stress levels appear to vary according to sleep disorder status. Participants with no sleep disorder generally exhibit lower stress levels, with most observations between 4 and 6. In contrast, individuals with insomnia tend to report moderately higher stress levels, while those with sleep apnea show the highest overall stress distribution and the greatest variability. The wider interquartile range among participants with sleep apnea suggests that stress experiences differ considerably within this group. Overall, the visualization suggests a positive relationship between sleep disorders and increased stress levels.

#### Interpretation of Physical Activity Level by Sleep Disorder

The boxplot suggests clear differences in physical activity levels among the three sleep disorder categories. Participants with sleep apnea generally demonstrate the highest physical activity levels, with a median close to 75. Individuals without a sleep disorder display moderate activity levels, although considerable variability exists within this group. Participants with insomnia tend to have the lowest average physical activity levels, with observations concentrated around 45. The presence of jittered observations indicates that while variability exists within each category, the overall differences between the groups remain apparent.

#### Overall Interpretation

The comparative boxplots suggest that sleep disorder status is associated with meaningful differences in several health indicators. Participants without a sleep disorder generally experience better sleep quality, longer sleep duration, lower stress levels, and moderate physical activity. Individuals with insomnia consistently exhibit poorer sleep quality, shorter sleep duration, elevated stress levels, and comparatively lower physical activity. Participants with sleep apnea demonstrate the greatest variability across the measured health indicators, particularly in sleep quality, sleep duration, and stress level, suggesting that the condition affects individuals differently. Overall, these findings indicate that sleep disorders are associated with changes in multiple aspects of health and well-being, highlighting the importance of considering sleep disorder status when evaluating sleep and lifestyle outcomes.

- Does longer sleep reduce stress?

### 7.6 Correlation Analysis

Correlation analysis measures the strength and direction of linear relationships between numerical variables.

The Pearson correlation coefficient ranges from:

- +1 → Perfect positive correlation
- 0 → No linear correlation
- -1 → Perfect negative correlation

A heatmap is used to visualize the correlation matrix, making it easier to identify variables that are strongly related

The **Person ID** column was excluded from the correlation analysis because it is an identifier rather than a measurable health variable. Including identifier variables in a correlation matrix can produce misleading relationships that have no practical interpretation. Therefore, only meaningful numerical health indicators were retained for the analysis.

```python
# Select only numerical variables
numerical_data = df_sleep.select_dtypes(include=['int64', 'float64'])

# Remove Person ID since it is an identifier
numerical_data = numerical_data.drop(columns=['Person ID'])

# Calculate the correlation matrix
correlation_matrix = numerical_data.corr()

# Create the heatmap
plt.figure(figsize=(10, 8))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5,
    square=True
)

plt.title(
    'Correlation Matrix of Numerical Variables',
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()
plt.show()
```

#### Interpretation of the Correlation Heatmap

The correlation heatmap reveals several important relationships among the numerical variables in the Sleep and Health dataset.

The strongest positive correlation is observed between Sleep Duration and Quality of Sleep (r = 0.88). This indicates that participants who sleep for longer durations generally report better sleep quality. This strong relationship supports the trend observed in the scatter plot.

A strong positive relationship is also found between Physical Activity Level and Daily Steps (r = 0.77). This suggests that participants who engage in higher levels of physical activity also tend to record more daily steps, indicating consistency between these two measures of activity.

Conversely, Quality of Sleep and Stress Level exhibit a very strong negative correlation (r = -0.90). This indicates that higher stress levels are strongly associated with poorer sleep quality. Similarly, Sleep Duration is strongly negatively correlated with Stress Level (r = -0.81), suggesting that individuals experiencing greater stress generally sleep for fewer hours.

Heart rate demonstrates a moderate positive relationship with Stress Level (r = 0.67), implying that participants with higher stress levels tend to have higher resting heart rates. In contrast, heart rate shows moderate negative correlations with Quality of Sleep (r = -0.66) and Sleep Duration (r = -0.52), indicating that better sleep is generally associated with lower heart rates.

Age exhibits only weak to moderate relationships with the other variables. It has a moderate positive correlation with Quality of Sleep (r = 0.47) and a weak positive correlation with Sleep Duration (r = 0.34). Age also shows weak negative relationships with Stress Level (r = -0.42) and Heart Rate (r = -0.23), suggesting that age has a relatively limited influence on these health indicators within this dataset.

Most remaining variable pairs have correlation coefficients close to zero, indicating little or no linear relationship. For example, Daily Steps has almost no correlation with Sleep Duration (r = -0.04), Quality of Sleep (r = 0.02), and Heart Rate (r = -0.03).

Overall, the correlation analysis indicates that sleep duration, sleep quality, and stress level are the most closely related health indicators in the dataset. The findings suggest that individuals who sleep longer tend to experience better sleep quality and lower stress levels, while increased stress is associated with poorer sleep outcomes and higher heart rates.
